```python
class Chocolate:
  def __init__(self, price, adder = None):
    self.price = price
    self.adder = adder
    type(price) is float
    type(adder) is str
    if price < 0:
      raise ValueError('Цена должна быть больше нуля')
  def show_my_chocolate(self):
    if self.adder is None:
     print('Обычный шоколадный батончик')
    else:
     print(f'Шоколадный батончик, добавка:{self.adder}')
  def __str__(self):
    return(f'Шоколадный батончик, {self.price} рублей')
```

```python
nuts = Chocolate(15.5, 'Корица')
nuts.show_my_chocolate()
print(nuts)
```

```python
nuts = Chocolate(-15.5, 'Корица')
nuts.show_my_chocolate()
print(nuts)
```

```python
nuts = Chocolate(15.5)
nuts.show_my_chocolate()
print(nuts)
```
